```python
# 通过apply对Series中的数据进行处理

import pandas as pd

def func(item):
    # 注意：如果是在Series上应用apply，自定义函数接收的参数是一个具体的元素对象
    if item == 10:
        return item * 2
    elif item == 20:
        return item * 3
    else:
        return item * 4

s = pd.Series([10,20,30])

s.apply(func)

s.apply(lambda item: item * 2)

# 在调用函数的时候，可以通过关键字的形式进行传参
def func(item,p):
    return item * p

s.apply(func,p=2)

```

```python
#DataFrame使用apply()

def func(item):
    # 注意：如果是在DataFrame上应用apply，自定义函数接收的参数是一个Series对象
    return item.sum()

df = pd.DataFrame({"a": [10, 20, 30], "b": [40, 50, 60]})
# df.sum()
df.apply(func,axis=0)
```

```python
def func(s):
    return s["a"] / s["b"]

df = pd.DataFrame({"a": [10, 20, 30], "b": [40, 50, 60]})
df.apply(func,axis=1)
```

```python
# 向量化函数
import numpy as np

@np.vectorize
def f(x, y):
    if y == 0:
        return np.nan
    return x / y

df = pd.DataFrame({"a": [10, 20, 30], "b": [40, 0, 60]})

f(df["a"], df["b"])
# 将函数转换为向量化函数
# f_vec = np.vectorize(f)
# f_vec(df["a"], df["b"])
```
