```python
import pandas as pd
from pandas import DataFrame

# Series和Series连接
s1 = pd.Series(["A", "B"], index=[1, 2])
s2 = pd.Series(["D", "E"], index=[4, 5])
s3 = pd.Series(["G", "H"], index=[7, 8])

pd.concat([s1, s2, s3])
pd.concat([s1, s2, s3],axis=1)


```

```python
# DataFrame和Series连接
df1 = pd.DataFrame(data={"a": [1, 2], "b": [4, 5]}, index=[1, 2])
s1 = pd.Series(data=[7, 10], index=[1, 2], name="a")

s1

pd.concat([df1,s1])
pd.concat([df1,s1],axis=1)
```

```python
# DataFrame与DataFrame连接
df1 = pd.DataFrame(data={"a": [1, 2], "b": [4, 5]}, index=[1, 2])
df2 = pd.DataFrame(data={"a": [7, 8], "b": [10, 11]}, index=[1, 2])

df2
pd.concat([df1,df2])
pd.concat([df1,df2],axis=1)
```

```python
df1 = pd.DataFrame(data={"a": [1, 2], "b": [4, 5]}, index=[1, 2])
df2 = pd.DataFrame(data={"a": [7, 8], "b": [10, 11]}, index=[1, 2])

df2
# 通过ignore_index=True重置索引
pd.concat([df1,df2],ignore_index=True)
```

```python
# 通过concat实现类似join的效果
df1 = pd.DataFrame(data={"a": [1, 2], "b": [4, 5]}, index=[1, 2])
df2 = pd.DataFrame(data={"b": [7, 8], "c": [10, 11]}, index=[2, 3])

df2
# 默认实现的是类似全外连接的效果
pd.concat([df1,df2],join="outer")

pd.concat([df1,df2],join="inner",axis=1)
# pd.concat([df1,df2],join="left",axis=1)

# 总结：concat一般就是在对结果集进行合并的时候使用

```
