```python
import numpy as np

arr1 = np.array([[1.3,-2.6,3.5],[4.7,-5.1,6.5]])
arr1
# np.abs()	元素的绝对值，参数是 number 或 array
np.abs(arr1)
# np.ceil()	向上取整，参数是 number 或 array
np.ceil(arr1)

# np.floor()	向下取整，参数是 number 或 array
np.floor(arr1)
# np.rint()	四舍五入，参数是 number 或 array
# 四舍六入：当需要舍入的数字小于 5 时，直接舍去；当需要舍入的数字大于 5 时，进位。
# 五成双：当需要舍入的数字恰好是 5 时，会看 5 前面的数字，如果是偶数则直接舍去 5，如果是奇数则进位。
np.rint(arr1)


arr2 = np.array([[1.3,np.nan,3.5],[4.7,np.nan,6.5]])
arr2
# print(arr2.dtype)
np.isnan(arr2)
# type(np.nan)
# np.isnan()	判断元素是否为NaN(Not a Number) ，参数是 number 或 array

# np.multiply()	元素相乘，参数是 number 或 array。如果第二个参数传递的是number，原数组中所有元素乘以这个数字，返回新的数组；
# 如果第二个参数也是一个数组，是将两个数组中对应位置的元素相乘，返回一个新的数组，其形状与输入数组相同。
arr3 = np.array([[1,2,3],[4,5,6]])
arr4 = np.array([[10,20,30],[40,50,60]])

# arr4
np.multiply(arr3,arr4)
# np.divide()	元素相除，参数是 number 或 array
np.divide(arr4,arr3)

# np.where(condition, x, y)	三元运算符，x if condition else y
np.where(arr3 > 4 ,11,22)

```
