```python
import pandas as pd

arrs = pd.Series([11,22,33,44,55],name="atguigu",index=["a","b","c","d","e"])
arrs
# Series常用的属性
# index	Series的索引对象
arrs.index
# values	Series的值
arrs.values
type(arrs.values)
# ndim	Series的维度
arrs.ndim
# shape	Series的形状
arrs.shape
# size	Series的元素个数
arrs.size
# dtype或dtypes	Series的元素类型
arrs.dtypes
# name	Series的名称
arrs.name


# loc[]	显式索引，按标签索引或切片
arrs.loc["a":"c"]
# iloc[]	隐式索引，按位置索引或切片
arrs.iloc[0:2]

# at[]	使用标签访问单个元素
arrs.at["c"]
# iat[]	使用位置访问单个元素
arrs.iat[2]
```
