```python
# 该案例演示了Series对象的创建

import pandas as pd

# 通过列表创建Series
s1 = pd.Series([3,1,2,4])

# 在创建Series对象的时候，显示的指定标签
s2 = pd.Series([3,1,2,4,5],index=['a','b','c','d','e'])
s2

# 在创建Series对象的时候，显示的指定标签以及名字
s3 = pd.Series([3,1,2,4,5],index=['a','b','c','d','e'],name='numbers')
s3

# 通过字典的方式创建Series对象
s4 = pd.Series({"a":1,"b":2,"c":3,"d":4,"e":5},name="numbers")
s4

dic = {"a": 4, "b": 7, "c": -5, "d": 3}

s5 = pd.Series(dic, name="numbers")

s6 = pd.Series(dic, index=["a","d"], name="numbers")
s6











```
