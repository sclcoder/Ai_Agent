```python
# Series常用的方法
import numpy as np
import pandas as pd
arrs = pd.Series([11,22,np.nan,None,44,22],index=['a','b','c','d','e','f'])
arrs

# head()	查看前n行数据，默认5行
arrs.head(2)
# tail()	查看后n行数据，默认5行
arrs.tail(2)

# isin()	元素是否包含在参数集合中
arrs.isin([11,44,np.nan])

# isna()	元素是否为缺失值（通常为 NaN 或 None）
arrs.isna()

# sum()	求和，会忽略 Series 中的缺失值
arrs.sum()
# mean()	平均值
arrs.mean()
# min()	最小值
arrs.min()
# max()	最大值
arrs.max()
# var()	方差
arrs.var()
# std()	标准差
arrs.std()
# median()	中位数
arrs.median()
# mode()	众数（出现频率最高的值），如果有多个值出现的频率相同且都是最高频率，这些值都会被包含在返回的 Series 中
arrs.mode()

arrs.quantile(0.25,interpolation='midpoint')
# quantile(q,interpolation)	指定位置的分位数
# q的取值范围是 0 到 1 之间的浮点数或浮点数列表，如quantile(0.5)表示计算中位数（即第 50 百分位数）;
# interpolation：指定在计算分位数时，如果分位数位置不在数据点上，采用的插值方法。默认值是线性插值 'linear'，还有其他可选值如 'lower'、'higher'、'midpoint'、'nearest' 等


# describe()	常见统计信息
arrs.describe()

# value_count()	每个元素的个数
type(arrs.value_counts())

# count()	非缺失值元素的个数，如果要包含缺失值，用len()
len(arrs)
arrs.count()
arrs.size

# drop_duplicates()	去重
arrs.drop_duplicates()

# unique()	去重后的数组
type(arrs.unique())


# nunique()	去重后元素个数
arrs.nunique()

# sample()	随机采样
arrs.sample()

arrs = pd.Series([11,22,np.nan,None,44,22],index=['d','e','a','c','f','b'])
# sort_index()	按索引排序
arrs.sort_index()
# sort_values()	按值排序
arrs.sort_values(ascending=False)

# replace()	用指定值代替原有值
arrs.replace(22,66)

# to_frame()	将Series转换为DataFrame
type(arrs.to_frame())
# equals()	判断两个Series是否相同
arr1 = pd.Series([1,2,3])
arr2 = pd.Series([1,2,3,4])
arr1.equals(arr2)

# keys()	返回Series的索引对象
arrs.index
arrs.keys()
# corr()	计算与另一个Series的相关系数
# 默认使用皮尔逊相关系数（Pearson correlation coefficient）来计算相关性。要求参与比较的数组元素类型都是数值型。
# 当相关系数为 1 时，表示两个变量完全正相关，即一个变量增加，另一个变量也随之增加。
# 当相关系数为 -1 时，表示两个变量完全负相关，即一个变量增加，另一个变量随之减少。
# 当相关系数为 0 时，表示两个变量之间不存在线性相关性。
# 例如，分析某地区的气温和冰淇淋销量之间的关系
arr1 = pd.Series([1,2,3])
arr2 = pd.Series([1,2,3])

arr3 = pd.Series([3,2,1])
arr4 = pd.Series([6,7,8])
arr5 = pd.Series([1, -1, 1, -1])
arr6 = pd.Series([1, 1, -1, -1])

# print(arr1.corr(arr2))
# print(arr1.corr(arr3))
# print(arr1.corr(arr4))
# print(arr5.corr(arr6))


# cov()	计算与另一个Series的协方差
# print(arr1.cov(arr3))

# hist()	绘制直方图，用于展示数据的分布情况。它将数据划分为若干个区间（也称为 “bins”），并统计每个区间内数据的频数。需要安装matplotlib包
arr7 = pd.Series([3,2,1,1,1,2,2])
# arr7.hist(bins=5)

# items()	获取索引名以及值
for i,v in arrs.items():
    print(i,v)

```

```python
# 布尔索引
s = pd.Series({"a": -1.2, "b": 3.5, "c": 6.8, "d": 2.9})

bools = s > s.mean()
bools
s[bools]
```

```python
# Series的运算
# 1）Series与标量运算 标量会与每个元素进行计算
s = pd.Series({"a": -1.2, "b": 3.5, "c": 6.8, "d": 2.9})
s * 10

# 2）Series与Series运算 会根据标签索引进行对位计算，索引没有匹配上的会用NaN填充。
s = pd.Series([1,2,3,4])
s1 = pd.Series([10,20,30,40],index=[1,2,3,4])
s + s1


```
