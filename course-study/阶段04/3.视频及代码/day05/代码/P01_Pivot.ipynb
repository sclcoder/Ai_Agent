```python
import pandas as pd

# 加载数据
df = pd.read_csv("E:\\250716_大模型\\05_尚硅谷大模型技术之Numpy&Pandas\\2.资料\\data\\sleep.csv")
df

# 统计不同睡眠时间，不同压力等级下的睡眠质量
# df.groupby(["sleep_duration",'stress_level'])["sleep_quality"].mean()

# 对睡眠时间进行划分
sleep_duration_stage =  pd.cut(df["sleep_duration"],[0, 5, 6, 7, 8, 9, 10, 11, 12])

# 对压力等级进行划分
stress_level_stage = pd.cut(df["stress_level"],5)

# 统计不同睡眠时间，不同压力等级下的睡眠质量
# df.groupby([sleep_duration_stage,stress_level_stage,"occupation"],observed=True)["sleep_quality"].mean().unstack()

# 通过透视表实现
# df.pivot_table(values="sleep_quality",index=[sleep_duration_stage,stress_level_stage],aggfunc="mean",observed=True)
#
# 添加职业作为列维度
df.pivot_table(
    values="sleep_quality",
    index=[sleep_duration_stage,stress_level_stage],
    columns=["occupation"],
    aggfunc="mean",
    observed=True)

# 添加性别作为第二个列维度
df.pivot_table(
    values="sleep_quality",
    index=[sleep_duration_stage,stress_level_stage],
    columns=["occupation","gender"],
    aggfunc="mean",
    observed=True)

df.groupby([sleep_duration_stage,stress_level_stage,"gender","occupation"],observed=True)["sleep_quality"].mean().unstack().unstack()
```
