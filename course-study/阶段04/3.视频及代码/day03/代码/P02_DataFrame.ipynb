# DataFrame的创建

```python
import pandas as pd

mongo_series = pd.Series([4,5,6,3,1],name='Mango')
apple_series = pd.Series([5,4,3,0,2],name='Apple')
banana_series = pd.Series([2,3,5,2,7],name='Banana')

df = pd.DataFrame({"Mango":mongo_series,"Apple":apple_series,"Banana":banana_series})
df


df = pd.DataFrame({"id":[100,200,300],"name":["张三","李四","王五"],"age":[10,20,30]})
df

# 在创建对象的时候，指定行标签以及列的顺序
df = pd.DataFrame({"name":["张三","李四","王五"],"age":[10,20,30]},columns=["age","name"],index=[100,200,300])
df
```

## DataFrame常用的属性

```python
import pandas as pd
df = pd.DataFrame(data={"id": [101, 102, 103], "name": ["张三", "李四", "王五"], "age": [20, 30, 40]},index=["aa", "bb", "cc"])

df
# index	DataFrame的行索引
df.index
# columns	DataFrame的列标签
df.columns

# values	DataFrame的值
df.values
# ndim	DataFrame的维度
df.ndim
# shape	DataFrame的形状
df.shape
# size	DataFrame的元素个数
df.size
# dtypes	DataFrame的元素类型
df.dtypes
# T	行列转置
df.T

# loc[]	显式索引，按行列标签索引或切片
df.loc["aa","id":"age"]
# iloc[]	隐式索引，按行列位置索引或切片
df.iloc[1:,:]
# at[]	使用行列标签访问单个元素
df.at["bb","name"]
# iat[]	使用行列位置访问单个元素
df.iat[1,1]
```

## DataFrame常用方法

```python
import pandas as pd
import numpy as np
df = pd.DataFrame(data={"id": [101, 102, 103,104,105,106,107], "name": ["张三", "李四", "王五","赵六","冯七","周八","张三"], "age": [10, 20, 30, 40, None, 60,10]},index=["aa", "bb", "cc", "dd", "ee", "ff","aa"])
df
# head()	查看前n行数据，默认5行
df.head()
# tail()	查看后n行数据，默认5行
df.tail(3)
# isin()	元素是否包含在参数集合中
df.isin([101,"李四",np.nan])
# isna()	元素是否为缺失值
df.isna()
# sum()	求和
df["age"].sum()

# mean()	平均值
df["age"].mean()
# min()	最小值
df["age"].min()
# max()	最大值
df["age"].max()
# var()	方差
df["age"].var()
# std()	标准差
df["age"].std()
# median()	中位数
df["age"].median()
# mode()	众数
df["age"].mode()
# quantile()	指定位置的分位数，如quantile(0.5)
df["age"].quantile()

# describe()	常见统计信息
df.describe(include=[np.float64])
# info()	基本信息
# df.info()
# value_counts()	每个元素的个数
df.value_counts()

# count()	非空元素的个数
df.count()
# drop_duplicates()	去重
df.drop_duplicates()
# sample()	随机采样
df.sample()

# replace()	用指定值代替原有值
# df.replace(np.nan, 88)
# equals()	判断两个DataFrame是否相同
df1 = pd.DataFrame(data={"id": [101, 102, 103], "name": ["张三", "李四", "王五"], "age": [10, 20, 30]})
df2 = pd.DataFrame(data={"id": [101, 102, 104], "name": ["张三", "李四", "王五"], "age": [10, 20, 30]})
# print(df1.equals(df2))

df3 = pd.DataFrame({'A': [2, 5, 3, 7, 4],'B': [1, 6, 2, 8, 3]})
# cummax()	累计最大值
df3.cummax(axis=1)
# cummin()	累计最小值
df3.cummin(axis=1)
# cumsum()	累计和
df3.cumsum(axis=1)
# cumprod()	累计积
df3.cumprod()
# diff()	一阶差分，对序列中的元素进行差分运算，也就是用当前元素减去前一个元素得到差值，默认情况下，它会计算一阶差分，即相邻元素之间的差值。参数：
# periods：整数，默认为 1。表示要向前或向后移动的周期数，用于计算差值。正数表示向前移动，负数表示向后移动。
# axis：指定计算的轴方向。0 或 'index' 表示按列计算，1 或 'columns' 表示按行计算，默认值为 0。

# df3.diff()
# df3.diff(axis=1)
df3.diff(periods=2)


# sort_index()	按行索引排序
df.sort_index(ascending=False,inplace=True)
df

# sort_values()	按某列的值排序，可传入列表来按多列排序，并通过ascending参数设置升序或降序
df.sort_values(by=["age","id"],ascending=[True,False])

# nlargest()	返回某列最大的n条数据
df.nlargest(3,columns=["age"])
# nsmallest()	返回某列最小的n条数据
df.nsmallest(3,columns=["age"])

```


## 布尔索引

```python
df = pd.DataFrame(
    data={"age": [20, 30, 40, 10], "name": ["张三", "李四", "王五", "赵六"]},
    columns=["name", "age"],
    index=[101, 104, 103, 102],
)
bools = df["age"] > 25

df[bools]


df[df["age"] > 25]
```

## DataFrame的运算

```python
# DataFrame和标量进行运算    标量会和DataFrame的每一个元素进行运算
df = pd.DataFrame(
    data={"age": [20, 30, 40, 10], "name": ["张三", "李四", "王五", "赵六"]},
    columns=["name", "age"],
    index=[101, 104, 103, 102],
)
df * 2
df["age"] * 2

# 2）DataFrame与DataFrame运算
# 根据标签索引进行对位计算，索引没有匹配上的用NaN填充

df1 = pd.DataFrame(
    data={"age": [10, 20, 30, 40], "name": ["张三", "李四", "王五", "赵六"]},
    columns=["name", "age"],
    index=[101, 102, 103, 104],
)
df2 = pd.DataFrame(
    data={"age": [10, 20, 30, 40], "name": ["张三", "李四", "王五", "田七"]},
    columns=["name", "age"],
    index=[102, 103, 104, 105],
)
df2

df1 + df2
```

## DataFrame的更改操作

```python
# 设置行索引(标签)
df = pd.DataFrame({"age": [20, 30, 40, 10], "name": ["张三", "李四", "王五", "赵六"], "id": [101, 102, 103, 104]})
df
# 通过set_index()设置行索引   将已有列设置为行标签
# df1 = df.set_index("id")
# df1.columns
# df1.index
df.set_index("id",inplace=True)
# df.columns
# df.index
# 通过reset_index重置行索引
df.reset_index(inplace=True)



```

```python

df = pd.DataFrame({"age": [20, 30, 40, 10], "name": ["张三", "李四", "王五", "赵六"], "id": [101, 102, 103, 104]})
df
df.set_index("id", inplace=True)
# 修改行索引名  以及列名
df.rename(index={101: "一", 102: "二", 103: "三", 104: "四"}, columns={"age": "年龄", "name": "姓名"}, inplace=True)
df
# 通过重新给index和column属性 赋值的方式修改名字
df.index = ["Ⅰ", "Ⅱ", "Ⅲ", "Ⅳ"]
df.columns = ["年齡", "名稱"]
df
```
