```python
# 该案例演示了numpy中ndarray数组相关的属性

# 导入numpy模块
import numpy as np

# 创建一个ndarray数组对象
# a = [[1,2,3],[4,5,6]]
# print(type(a))

arr = np.array([[1,2,3],[4,5,6]])
# print(type(arr))
# print(arr)
arr
# 维度
print(arr.ndim)
# 形状
print(arr.shape)
# 元素个数
print(arr.size)
# 元素数据类型
print(arr.dtype)
# 每一个元素字节大小
print(arr.itemsize)
```
