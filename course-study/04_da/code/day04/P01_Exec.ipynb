1. 创建一个包含'姓名'和'年龄'两列的DataFrame，数据如下：

| 姓名 | 年龄 |

|----|----|

|Alice|25|

|Bob|30|

|Charlie|35|

然后计算所有人员的平均年龄。---获取年龄字段 df['年龄']或者df.年龄

```python
import pandas as pd

df = pd.DataFrame({"姓名":["Alice","Bob","Charlie"],"年龄":[25,30,35]})
df["年龄"]
df.年龄.mean()
```

2. 构建一个DataFrame df如下：

|Fruit|Price|Quantity|

-----------------------------------------

|Apple|2|5|

|Banana|1|8|

|Apple|3|3|

|Banana|1.5|6|

使用groupby函数按Fruit分组，计算每种水果的总销售额（销售额 = 价格 * 数量）。

```python
import pandas as pd
df = pd.DataFrame({"Fruit":["Apple","Banana","Apple","Banana"],"Price":[2,1,3,1.5],"Quantity":[5,8,3,6]})
df["total"] = df["Price"] * df["Quantity"]
df.groupby("Fruit")["total"].sum()
```
