```python
from doctest import DocFileCase

# 一对一
import pandas as pd
df1 = pd.DataFrame(
    {"employee": ["Bob", "Jake", "Lisa", "Sue"], "group": ["Accounting", "Engineering", "Engineering", "HR"]}
)
df2 = pd.DataFrame({"employee": ["Lisa", "Bob", "Jake", "Sue"], "hire_date": [2004, 2008, 2012, 2014]})

df2

# 默认用两个df中同名字段进行连接
pd.merge(df1, df2)
df1.merge(df2)
```

```python
# 多对一
df1 = pd.DataFrame(
    {"employee": ["Bob", "Jake", "Lisa", "Sue"], "group": ["Accounting", "Engineering", "Engineering", "HR"]}
)
df2 = pd.DataFrame({"group": ["Accounting", "Engineering", "HR"], "supervisor": ["Carly", "Guido", "Steve"]})


pd.merge(df1, df2)
```

```python
# 多对多
df1 = pd.DataFrame(
    {"employee": ["Bob", "Jake", "Lisa", "Sue"], "group": ["Accounting", "Engineering", "Engineering", "HR"]}
)
df2 = pd.DataFrame(
    {
        "group": ["Accounting", "Accounting", "Engineering", "Engineering", "HR", "HR"],
        "skills": ["math", "spreadsheets", "coding", "linux", "spreadsheets", "organization"],
    }
)

pd.merge(df1, df2)
```

```python
# （1）通过on指定使用某个列连接，只能在有共同列名的时候使用
df1 = pd.DataFrame(
    {"employee": ["Bob", "Jake", "Lisa", "Sue"], "group": ["Accounting", "Engineering", "Engineering", "HR"]}
)
df2 = pd.DataFrame({"employee": ["Lisa", "Bob", "Jake", "Sue"], "hire_date": [2004, 2008, 2012, 2014]})

pd.merge(df1, df2,on="employee")
```

```python
# （2）两对象列名不同，通过left_on和right_on分别指定列名
df1 = pd.DataFrame(
    {"employee": ["Bob", "Jake", "Lisa", "Sue"], "group": ["Accounting", "Engineering", "Engineering", "HR"]}
)
df2 = pd.DataFrame({"name": ["Bob", "Jake", "Lisa", "Sue"], "salary": [70000, 80000, 120000, 90000]})

pd.merge(df1, df2,left_on="employee", right_on="name")


```

```python
# （3）通过left_index和right_index设置合并的索引
# 通过设置merge()中的left_index、right_index参数将索引设置为键来实现合并

df1 = pd.DataFrame(
    {"employee": ["Bob", "Jake", "Lisa", "Sue"], "group": ["Accounting", "Engineering", "Engineering", "HR"]}
)
df2 = pd.DataFrame({"name": ["Lisa", "Bob", "Jake", "Sue"], "hire_date": [2004, 2008, 2012, 2014]})


# pd.merge(df1,df2)
# pd.merge(df1,df2,on="employee")
# pd.merge(df1,df2,left_on="employee", right_on="employee")

df1.set_index("employee", inplace=True)
df2.set_index("name", inplace=True)

pd.merge(df1, df2,left_index=True, right_index=True)

# df2
```

```python
df1 = pd.DataFrame({"name": ["Peter", "Paul", "Mary"], "food": ["fish", "beans", "bread"]}, columns=["name", "food"])
df2 = pd.DataFrame({"name": ["Mary", "Joseph"], "drink": ["wine", "beer"]}, columns=["name", "drink"])

df2

# 默认使用的是内连接   只会将完全匹配的数据关联上
pd.merge(df1, df2)

# 可以通过how参数指定连接类型
# 默认how='inner'
pd.merge(df1, df2,how="inner")

# 全外连接
pd.merge(df1, df2, how="outer")

# concat只是对结果集进行合并，如果要实现全外连接效果，还是需要通过merge，指定how='outer'
# pd.concat([df1, df2])

# 左外连接
pd.merge(df1, df2, how="left")

# 右外连接
pd.merge(df1, df2, how="right")
#
# pd.merge(df1, df2, how="outer", on="name")

```

```python
# 4）重复列名的处理
df1 = pd.DataFrame({"name": ["Bob", "Jake", "Lisa", "Sue"], "rank": [1, 2, 3, 4]})
df2 = pd.DataFrame({"name": ["Bob", "Jake", "Lisa", "Sue"], "rank": [3, 1, 4, 2]})

 # 不指定后缀名，默认为_x和_y
# pd.merge(df1, df2,on="name")
pd.merge(df1, df2,on="name",suffixes=("_left", "_right"))

```
