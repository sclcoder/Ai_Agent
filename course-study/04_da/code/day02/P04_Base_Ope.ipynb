```python
# 该案例演示了np的基本运算
import numpy as np

# 大小相等的数组之间的任何算术运算都会将运算应用到元素级 对位计算
arr1 = np.array([[1, 2, 3], [4, 5, 6]])
arr2 = np.array([[7, 8, 9], [10, 11, 12]])
# arr2
# print(arr1 + arr2)
# print(arr1 - arr2)
# print(arr1 * arr2)
# print(arr1 / arr2)

# 数组与标量的算术运算会将标量值传播到各个元素
# arr1 + 5
```

```python
# 不同大小的数组之间的运算叫做广播

# 	规则1：如果俩个数组的维度数不相同，那么小维度数组的形状将会在最左边补1
# arr1 = np.array([1, 2, 3])  # 形状为 (3,)
# 二维数组
arr2 = np.array([[4], [5], [6]])  # 形状为 (3, 1)

arr1 = np.array([[1, 2, 3]])  # 形状为 (1,3)

# 	规则2：如果俩个数组的形状在任何一个维度上都不匹配，那么数组的形状会沿着维度大小（元素个数）为1的维度开始扩展 ，
# （维度必须是1开始）直到所有维度都一样， 以匹配另一个数组的形状。
arr1 + arr2


# 	规则3：如果俩个数组的形状在任何一个维度上都不匹配，并且没有任何一个维度大小等于1，那么会引发异常。

arr5 = np.array([1, 2, 3])  #(3,)
# arr6 = np.array([4, 5])     #(2,)
arr6 = np.array([4])     #(1,)
arr5 + arr6

# arr5 + 10
```
