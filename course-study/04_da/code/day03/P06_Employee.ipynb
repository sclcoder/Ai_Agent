## 员工统计分析

```python
import pandas as pd

# 加载员工数据集
df = pd.read_csv("../data/employees.csv")
df

df.nsmallest(1,"salary")
df.nlargest(1,"salary")
# 找出最低薪资的员工
df[df["salary"] == df["salary"].min()]
# 找出最低薪资的员工
df.loc[df["salary"] == df["salary"].min()]
# 找出最高薪资的员工
df.loc[df["salary"] == df["salary"].max()]

# 使用排序的方法找出最低薪资的员工
df.sort_values(by="salary", ascending=False).tail(1)
# 使用排序的方法找出最高薪资的员工
df.sort_values(by="salary", ascending=False).head(1)

df.nlargest(10, "salary")

df["department_id"].unique()

# 查看每个部门的员工数
# select did,count(*) from emp group did
df.groupby("department_id")["employee_id"].count()

df.groupby("department_id")["employee_id"].count().rename("employee_count").plot(kind="bar")

```
