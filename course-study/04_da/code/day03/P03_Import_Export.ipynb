DataFrame的导出和导入


```python
import os
import pandas as pd

os.makedirs("data", exist_ok=True)

df = pd.DataFrame({"age": [20, 30, 40, 10], "name": ["张三", "李四", "王五", "赵六"], "id": [101, 102, 103, 104]})
df.set_index("id", inplace=True)
df
# 将df数据导出到csv文件中
df.to_csv("data/test.csv",encoding="gbk")

# 将df数据导出到json
df.to_json("data/test.json",force_ascii=False)

# 导出到剪切板
df.to_clipboard()


```

```python
# 导出到mysql数据库
from sqlalchemy import create_engine


df = pd.DataFrame({"age": [20, 30, 40, 10], "name": ["张三", "李四", "王五", "赵六"], "id": [101, 102, 103, 104]})
df.set_index("id", inplace=True)

host="localhost"
port=3306
user="root"
passwd="123456"
db="atguigu"
charset="utf8mb4"

# 创建连接引擎
engine = create_engine( f"mysql+pymysql://{user}:{passwd}@{host}:{port}/{db}?charset={charset}")

df.to_sql("t_test", engine, if_exists="replace")
```

```python
# 从csv中导入数据
pd.read_csv("data/test.csv", encoding="gbk")

# 从json中导入数据
pd.read_json("data/test.json")

# 从mysql中导入数据
pd.read_sql("t_test", engine)
```
