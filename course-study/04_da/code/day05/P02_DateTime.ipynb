```python
from datetime import datetime

# python语言标准库中自带了  datetime日期类型
date1 = datetime(year=2000, month=1, day=1)
# type:datetime.datetime
date1

date2 = datetime.now()
date2
print(date1)  # 2000-01-01 00:00:00
print(date2)  # 2025-01-01 00:00:00
print(date1.year)  # 2000
print(date1.month)  # 1
print(date1.day)  # 1
print(date2.weekday())  # 5
print(date2.strftime("%A"))  # Saturday
print(date2 - date1)  # 18263 days, 0:00:00
```

```python
# 通过pandas的日期对象  提取日期各个部分
d = pd.Timestamp("2015-01-01 09:08:07.123456")
print(d.year)  # 2015
print(d.month)  # 1
print(d.day)  # 1
print(d.hour)  # 9
print(d.minute)  # 8
print(d.second)  # 7
print(d.microsecond)  # 123456
```

```python
# 如果是Series，需要获取时间属性访问器对象
df = pd.read_csv("../data/weather.csv")

df_date = pd.to_datetime(df["date"])
df["year"] = df_date.dt.year
df["month"] = df_date.dt.month
df["day"] = df_date.dt.day
# print(df[["date", "year", "month", "day"]].tail())

df["quarter"] = df_date.dt.to_period("Q")
# df.info()

df["dif"] = df_date - df_date[0]
df.info()
```

```python
#使用时间作为索引
df = pd.read_csv("../data/weather.csv")
df["ymd"] = pd.to_datetime(df["date"])
# df.set_index("ymd", inplace=True)
# df["jd"] = df["ymd"].dt.to_period("Q")
# df.set_index("jd", inplace=True)

df["diff"] = df["ymd"] - df["ymd"][0]
df.set_index("diff", inplace=True)
df.info()
```
