```python
import pandas as pd

df = pd.read_csv("../data/sleep.csv")


```

```python
# 使用柱状图展示不同睡眠时长的数量
pd.cut(df["sleep_duration"],[0, 5, 6, 7, 8, 9, 10, 11, 12]).value_counts().plot(kind="bar",color=["red", "green", "blue", "yellow", "cyan", "magenta", "black", "purple"])
# pd.cut(df["sleep_duration"],[0, 5, 6, 7, 8, 9, 10, 11, 12]).value_counts().plot.bar(color=["red", "green", "blue", "yellow", "cyan", "magenta", "black", "purple"])
```

```python
# 使用折线图展示不同睡眠时长的数量
# pd.cut(df["sleep_duration"], [0, 5, 6, 7, 8, 9, 10, 11, 12]).value_counts().sort_index().plot()
pd.cut(df["sleep_duration"], [0, 5, 6, 7, 8, 9, 10, 11, 12]).value_counts().sort_index().plot.line()
```

```python
# 使用面积图展示不同睡眠时长的数量
# pd.cut(df["sleep_duration"], [0, 5, 6, 7, 8, 9, 10, 11, 12]).value_counts().sort_index().plot.area()
pd.cut(df["sleep_duration"], [0, 5, 6, 7, 8, 9, 10, 11, 12]).value_counts().sort_index().plot(kind="area")
```

```python
# 使用直方图展示不同睡眠时长的数量
df["sleep_duration"].value_counts().plot.hist()
```

```python
# 使用饼状图展示不同睡眠时长的占比
pd.cut(df["sleep_duration"], [0, 5, 6, 7, 8, 9, 10, 11, 12]).value_counts().sort_index().plot.pie()
```

```python
# 绘制睡眠时间与睡眠质量的散点图
df.plot.scatter(x="sleep_duration", y="sleep_quality")
```

```python
df.plot.hexbin(x="sleep_duration", y="sleep_quality", gridsize=10)
```

```python
# 绘制睡眠时间与睡眠质量的堆叠图
df["sleep_quality_stage"] = pd.cut(df["sleep_quality"], range(11))
df["sleep_duration_stage"] = pd.cut(df["sleep_duration"], [0, 5, 6, 7, 8, 9, 10, 11, 12])
df_pivot_table = df.pivot_table(
    values="person_id", index="sleep_quality_stage", columns="sleep_duration_stage", aggfunc="count",observed=True
)
df_pivot_table.plot.bar(stacked=True)
```
